```python
print("Number checker")
number1 = int(input("Enter 1st number:"))
number2 = int(input("Enter 2nd number:"))

if number1 == number2:
  print(f"{number1} is equal to {number2}.")

elif number1 > number2:
   print(f"{number1} is greater than {number2}.")

elif number1 < number2:
  print(f"{number1} is less than {number2}.")
```

```python
# Divisibility Checker
print("This is divisibility checker for no. 5")

entry = int(input("Please enter a number:"))

if entry %5 == 0:
  print(f"{entry} is divisible by 5.")

else:
  print(f"{entry} is not divisible by 5.")
```

```python
#4
n = float(input("Enter a no. : "))

if n%3 == 0 and n%5 == 0 :
  print(f"{n} is a multiple of both 3 & 5")
else:
  print(f"{n} is not a multiple of both 3 & 5")


# Alternate Method :
n = float(input("Enter a no. : "))

if n%3 == 0 :
  if n%5 == 0 :
    print(f"{n} is a multiple of both 3 & 5")
else:
  print(f"{n} is not a multiple of both 3 & 5")
```

#While Loop

```python
password = 1234
user_entry = int(input("Please enter your password:"))

while user_entry != password:
  print("Incorrect password")
  user_entry = int(input("Please enter your password:"))
```

```python
# Thermostat

temp = int(input("Enter the room temp:"))

while temp <19:
  print("Heating is on")
  temp = int(input("Enter the room temp:"))

else :
  print("The room is warm enough. Heating is off.")

```
